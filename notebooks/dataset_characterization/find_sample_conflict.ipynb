```python
import pandas as pd
```

```python
# get mgnify samples list
mgnify_df = pd.read_csv('/home/kevin/Desktop/gut_microbiome/dataset/mgnify_gut/mgnify_raw/combined_taxonomy_abundance.csv')
mgnify_samples = mgnify_df["sample_id"].tolist()
mgnify_samples
```

```python
hmc_df = pd.read_csv('/home/kevin/Desktop/gut_microbiome/dataset/hmc/sample_metadata.tsv', sep='\t', index_col=0)
```

```python
srs_list = hmc_df.index.tolist()
srr_list = hmc_df["srr"].tolist()
```

```python
duplicated_sample_srs = []
duplicated_sample_srr = []
# srs
srs_set = set(srs_list)
# srr
srr_set = set(srr_list)
for mgnify_sample in mgnify_samples:
    if mgnify_sample in srs_set:
        duplicated_sample_srs.append(mgnify_sample)
    if mgnify_sample in srr_set:
        duplicated_sample_srr.append(mgnify_sample)
```
