```python
h5ad_path = '/scratch/kchen13/hmc_final_fixed/pretrain.h5ad'
adata = ad.read_h5ad(h5ad_path, backed='r')  # Use backed='r' for large files
print(adata)
```

## Matrix Stats (X)

```python
import numpy as np

for i in range(5):
    # Extract the row from the AnnData X matrix
    row = adata.X[i, :]
    if hasattr(row, "toarray"):  # Convert sparse matrix to dense if needed
        row = row.toarray().flatten()
    else:
        row = np.asarray(row).flatten()
    
    # Filter for non-zero elements
    nonzero_row = row[row != 0]
    nonzero_count = len(nonzero_row)
    
    if nonzero_count == 0:
        print(f"Row {i} -> All values are zero (Count: 0)")
        continue

    print(
        f"Row {i} -> "
        f"Non-zero Count: {nonzero_count:<4} | "
        f"Min: {nonzero_row.min():<5} | "
        f"Max: {nonzero_row.max():<5} | "
        f"Mean: {nonzero_row.mean():<8.2f} | "
        f"Median: {np.median(nonzero_row):<6.1f} | "
        f"Std: {nonzero_row.std():<.2f}"
    )
```
