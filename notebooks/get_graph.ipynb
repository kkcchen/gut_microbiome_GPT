```python
import matplotlib.pyplot as plt

def plot_dict_counts(d, title="Number of Tokens Trained on for Different Foundation Models"):
    """
    d: dict[str, int]
    Plots a bar chart of key → value on log scale.
    """
    keys = list(d.keys())
    values = list(d.values())

    plt.figure(figsize=(6, 4))
    colors = ['orange' if i == 0 else 'C0' for i in range(len(keys))]
    plt.bar(keys, values, color=colors)

    plt.yscale("log")  # ← log scale
    plt.xticks(rotation=45, ha="right")
    plt.ylabel("Number of Tokens (log scale)")
    plt.title(title)

    plt.tight_layout()
    plt.show()

```

```python
d = {
    'Microbiome Model' : 106469,
    'scGPT': 33e6,
    'GPT-1': 800e6,
    'GPT-3': 300e9
}

plot_dict_counts(d)
```
